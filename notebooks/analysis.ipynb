## A. Dataset Loading
We'll start by loading some sample data that mimics the structure of our scraped or uploaded resumes. We'll use `pandas` to easily manipulate and view this data.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt

# Add the parent directory to the path so we can seamlessly import our system modules anywhere
project_root = os.path.abspath('..')
if project_root not in sys.path:
    sys.path.append(project_root)

# Defining a representative JSON/CSV dataset snapshot
data = {
    "resume_id": [1, 2, 3, 4, 5],
    "text": [
        "Experienced software engineer with 5 years in Python, Machine Learning, and Data Analysis. Familiar with SQL and AWS.",
        "Frontend developer specialized in React, JavaScript, HTML, and CSS. Strong UI/UX design skills.",
        "Data Scientist matching business needs with Deep Learning and ML. Proficient in Python, TensorFlow, and SQL.",
        "Backend developer with experience in Node.js, Express, MongoDB, and Python.",
        "Machine Learning Engineer working with Computer Vision, PyTorch, and Python."
    ],
    "role_target": ["Backend Developer", "Frontend Developer", "Data Scientist", "Backend Developer", "ML Engineer"]
}
df = pd.DataFrame(data)
display(df.head())
```

## B. Data Exploration (EDA)
Let's explore our dataset, print out a specific sample, and generate a simple visual chart of the common target roles. This helps us ensure the data is rich and evenly distributed.

```python
print("--- Dataset Overview ---")
print(f"Total resumes loaded: {len(df)}")
print("\nSingle Sample Resume Text:")
print(f"'{df.iloc[0]['text']}'")

# Visualization: Target Roles frequency
role_counts = df['role_target'].value_counts()
plt.figure(figsize=(8, 5))
role_counts.plot(kind='bar', color='#4CAF50')
plt.title('Distribution of Target Roles in Dataset')
plt.xlabel('Role')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## C. Skill Extraction Testing
We simulate passing a resume through our `ai_engine.resume_parser` module to test how accurately we can pull unstructured information into standard skill sets.

```python
try:
    # Import our live engine module
    from ai_engine.resume_parser import extract_skills
    print("✅ Successfully imported extract_skills from ai_engine.resume_parser")
except (ImportError, ModuleNotFoundError):
    print("⚠️ Note: Mocking resume_parser to ensure demo continuity (module missing).")
    def extract_skills(text):
        # A simulated keyword extraction 
        keywords = ["Python", "Machine Learning", "Data Analysis", "SQL", "AWS", "React", "JavaScript", "HTML", "CSS", "Deep Learning", "TensorFlow", "Node.js", "MongoDB", "PyTorch"]
        return [kw for kw in keywords if kw.lower() in text.lower()]

# Test extraction on the first resume instance
sample_text = df.iloc[0]['text']
extracted_skills = extract_skills(sample_text)

print("\n--- Skill Extraction Results ---")
print(f"Original Resume Entry:\n\t{sample_text}")
print(f"\nExtracted Skills:\n\t{extracted_skills}")
```

## D. Semantic Matching Test
Our engine uses `models/embeddings/embedding_model.py` to compare skills. Even if spelling or acronyms differ ('ML' vs 'Machine Learning'), our semantic matcher connects them.

```python
try:
    from models.embeddings.embedding_model import get_similarity
    print("✅ Successfully imported get_similarity from models.embeddings.embedding_model")
except (ImportError, ModuleNotFoundError):
    print("⚠️ Note: Mocking embedding_model to ensure demo continuity (module missing).")
    def get_similarity(skill1, skill2):
        # Simulated semantic similarity map for embeddings
        sim_map = {
            ("ML", "Machine Learning"): 0.95,
            ("DL", "Deep Learning"): 0.94,
            ("Python", "Java"): 0.30,
            ("React", "Vue"): 0.85
        }
        return sim_map.get((skill1, skill2), sim_map.get((skill2, skill1), 0.5))

print("\n--- Semantic Similarity Representation ---")
print(f"Similarity ('ML' vs 'Machine Learning'): {get_similarity('ML', 'Machine Learning'):.2f}")
print(f"Similarity ('DL' vs 'Deep Learning'): {get_similarity('DL', 'Deep Learning'):.2f}")
print(f"Similarity ('Python' vs 'Java'): {get_similarity('Python', 'Java'):.2f}")
```

## E. Skill Gap Simulation
Now that we extracted the user's skills securely, we contrast them with the requirements of an arbitrary role defining our **Skill Gap**.

```python
try:
    from ai_engine.skill_gap import calculate_skill_gap
    print("✅ Successfully imported calculate_skill_gap from ai_engine.skill_gap")
except (ImportError, ModuleNotFoundError):
    print("⚠️ Note: Mocking skill_gap logic to ensure demo continuity (module missing).")
    def calculate_skill_gap(user_skills, required_skills):
        user_set = {s.lower() for s in user_skills}
        return [req for req in required_skills if req.lower() not in user_set]

user_skills = extracted_skills  # From Part C: ['Python', 'Machine Learning', 'Data Analysis', 'SQL', 'AWS']
role_required_skills = ["Python", "Machine Learning", "FastAPI", "Docker", "SQL", "Kubernetes"]

missing_skills = calculate_skill_gap(user_skills, role_required_skills)

print("\n--- Skill Gap Analysis ---")
print(f"User Has Skills: {user_skills}")
print(f"Role Required Skills: {role_required_skills}")
print(f"\n👉 Highlighted Skill Gap (Missing): {missing_skills}")
```

## F. Adaptive Learning Test
Using `ai_engine.adaptive_engine.py`, we translate the highlighted skill gap into actionable, highly targeted learning modules to close the gap.

```python
try:
    from ai_engine.adaptive_engine import generate_learning_path
    print("✅ Successfully imported generate_learning_path from ai_engine.adaptive_engine")
except (ImportError, ModuleNotFoundError):
    print("⚠️ Note: Mocking adaptive_engine generation to ensure demo continuity (module missing).")
    def generate_learning_path(missing_skills):
        path = {}
        for skill in missing_skills:
            path[skill] = f"Module: Introduction to {skill} | Est. Time: 2hrs"
        return path

learning_path = generate_learning_path(missing_skills)

print("\n--- Recommended Personal Adaptive Learning Path ---")
for skill, resource in learning_path.items():
    print(f"🔹 {skill} -> {resource}")
```

## G. Evaluation Metrics
We'll now evaluate the end-to-end performance using our `evaluation/metrics.py` module to quantitatively validate accuracy, gap reduction, and precision before and after applying the learning path.

```python
try:
    from evaluation.metrics import (
        skill_extraction_accuracy,
        skill_gap_reduction,
        recommendation_precision
    )
    print("✅ Successfully imported evaluation metrics")
    
    # 1. Pipeline Accuracy
    ground_truth_skills = ["Python", "Machine Learning", "Data Analysis", "SQL", "AWS", "Communication"]
    accuracy = skill_extraction_accuracy(extracted_skills, ground_truth_skills)
    
    # 2. Before vs After Gap Reduction (Assuming the user completed 2 out of 3 new modules)
    initial_gap = len(missing_skills)  # e.g., 3 (FastAPI, Docker, Kubernetes)
    final_gap = 1  # 2 modules completed
    gap_reduction = skill_gap_reduction(initial_gap, final_gap)
    
    # 3. Recommendation Precision check
    recommended_courses = list(learning_path.keys())
    relevant_to_role = ["FastAPI", "Docker", "Kubernetes"] 
    precision = recommendation_precision(recommended_courses, relevant_to_role)
    
    print("\n========== Final Metric Results ==========")
    print(f"📈 Skill Extraction Accuracy: {accuracy * 100:.1f}%")
    print(f"📉 Skill Gap Reduction:       {gap_reduction:.1f}% improvement")
    print(f"🎯 Recommendation Precision:  {precision * 100:.1f}%")
    print("=========================================")
    
except (ImportError, ModuleNotFoundError) as e:
    print(f"Could not import evaluation metrics: {e}")
```
